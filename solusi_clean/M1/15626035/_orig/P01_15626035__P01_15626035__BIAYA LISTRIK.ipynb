```python
data = {
    "daya": 0,
    "durasi": 0,
    "tarif": 0,
    "energi": 0,
    "biaya": 0}
```

```python
data["daya"] = int(input("Masukan daya lampu (watt):"))
data["durasi"] = int(input("Masukan durasi penggunaan lampu menyala (jam):"))
data["tarif"] = int(input("Masukan tarif listrik (rupiah per kWh):"))
```

```python
data["energi"] = data["daya"] * data["durasi"]
data["energi"] /= 1000
```

```python
data["biaya"] = data["energi"] * data["tarif"]
```

```python
print("Biaya listrik yang terbuang adalah Rp", int(data["biaya"]))
```
